```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

sys.path.insert(0, str(Path('..').resolve()))
from train_model import load_data, NUMERIC, CATEGORICAL, TARGET

df = load_data(Path('../data/dataset.csv'))
df.head()
```

```python
print(df.shape)
print('Missing values:', int(df.isna().sum().sum()))
print('Duplicate rows:', int(df.duplicated().sum()))
print('Months present:', sorted(df['Month'].unique()))
df[NUMERIC].describe().T
```

```python
counts = df[TARGET].value_counts().sort_index()
print(counts, '\nPurchase rate:', round(df[TARGET].mean(), 3))
print('Accuracy of always predicting no purchase:', round(1 - df[TARGET].mean(), 3))
counts.plot.bar(); plt.xticks([0, 1], ['No purchase', 'Purchase'], rotation=0); plt.title('Target distribution'); plt.show()
```

```python
print(df.groupby(TARGET)[NUMERIC].mean().T.round(2))
fig, ax = plt.subplots(1, 3, figsize=(14, 3.8))
for a, col in zip(ax, ['PageValues', 'ExitRates', 'ProductRelated_Duration']):
    df.boxplot(column=col, by=TARGET, ax=a, showfliers=False); a.set_title(col); a.set_xlabel('Revenue')
plt.suptitle(''); plt.tight_layout(); plt.show()
```

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 3.8))
for a, col in zip(ax, ['VisitorType', 'Month', 'Weekend']):
    df.groupby(col)[TARGET].mean().sort_values().plot.barh(ax=a); a.set_title(f'Purchase rate by {col}')
plt.tight_layout(); plt.show()
```

```python
corr = df[NUMERIC + [TARGET]].corr()
print(corr[TARGET].sort_values(ascending=False).round(2))
fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
ax.set_xticks(range(len(corr))); ax.set_xticklabels(corr.columns, rotation=75, ha='right')
ax.set_yticks(range(len(corr))); ax.set_yticklabels(corr.columns)
fig.colorbar(im); plt.tight_layout(); plt.show()
```
